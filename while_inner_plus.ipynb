```python
def get_innermost_plus_one(nested_list):
    current = nested_list
    while True:
        found_sublist = None
        for item in current:
            if isinstance(item, list):
                found_sublist = item
                break
        if found_sublist is not None:
            current = found_sublist
        else:
            break
    return [x +1 for x in current]

nested_data = [1, [2, 3, [4, 5]]]
print(get_innermost_plus_one(nested_data))
```
